```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt
```

```python
from google.colab import files
uploaded = files.upload()
```

```python
df = pd.read_csv("Tweets.csv")
df.head()
X = df["text"]
y = df["airline_sentiment"]
```

```python
vectorizer = CountVectorizer()

X_vec = vectorizer.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(X_vec,y,test_size=0.2)

model = MultinomialNB()
model.fit(X_train,y_train)

pred = model.predict(X_test)

print("Accuracy:",accuracy_score(y_test,pred))
```

```python
counts = df["airline_sentiment"].value_counts()

plt.figure()
counts.plot(kind='bar')

plt.title("Sentiment Distribution")
plt.xlabel("Sentiment")
plt.ylabel("Count")

plt.show()
```


**Sentiment Analysis Using Machine Learning**
*   The Naive Bayes model was trained on Twitter sentiment data.
*   Text data was converted into numerical features using CountVectorizer.
*   The model achieved an accuracy of around 76%, indicating good performance   in classifying positive and negative tweets.
